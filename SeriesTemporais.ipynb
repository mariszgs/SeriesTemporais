##O que são séries temporais?
Uma série temporal é um conjunto de observações coletadas ao longo de um determinado tempo, sendo o tempo um fator fundamental, pois a ordem em que os dados são observados pode influenciar a análise.

###Exemplos


*   Temperatura registrada ao longo dos dias
*   Vendas de uma empresa ao longo dos meses
*   Inflação ao longo dos anos

As observações podem ocorrer em intervalos regulares ou irregulares



```python
import pandas as pd

dados = {
    "Data": ["01/08/2026", "02/08/2026", "03/08/2026", "04/08/2026", "05/08/2026"],
    "Vendas": [120, 135, 90, 128, 163]
}

df = pd.DataFrame(dados)

df
```



Nesse exemplo, cada observação possui uma data e um número de vendas. A variável **Data** representa o tempo, enquanto **Vendas** representa o valor observado.
Como as vendas estão sendo acompanhadas ao longo do tempo, temos uma série temporal.

##Representação do Tempo

Os dados de uma série temporal podem representar o tempo de diferentes maneiras.

###Timestamp
Representa um instante específico no tempo.
**Exemplo:** 19/08/2026 às 17:38:00

###Período Fixo
Representa um período completo
**Exemplo:** Agosto de 2026

###Intervalo de Tempo
Representa um período definido por um início e um fim
**Exemplo:** de 01/08/2026 às 12:00 até 01/09/2026 às 12:00

###Tempo Decorrido
Representa o tempo passado desde o início de um evento ou experimento
**Exemplo:** Medir a temperatura de um forno a cada 10 segundos após o acionamento.

##Datetime e Timedelta

O python possui ferramentas próprias para trabalhar com datas, horários e intervalos de tempo.

### datetime
Representa uma data e um horário específicos.

### timedelta
Representa uma diferença ou duração de tempo.

Podemos utilizar essas ferramentas para realizar operações com datas, como calcular diferenças ou adicionar e remover períodos.

```python
from datetime import datetime, timedelta

data = datetime(2026, 8, 19, 17, 52)

data
```

```python
intervalo = timedelta(days=4)

nova_data = data + intervalo

nova_data
```

```python
data_inicial = datetime(2026, 8, 19)
data_final = datetime(2026, 8, 28)

diferenca = data_final - data_inicial

diferenca
```

Resumidamente:
- 'datetime' representa **quando** algo aconteceu
- 'timedelta' representa **quanto tempo** passou

Ambas são ferramentas base para trabalhar com datas antes de utilizar as ferramentas específicas do Pandas.

##Datas no Pandas

O pandas possui ferramentas específicas para trabalhar com conjuntos de datas.

Uma das principais é a função 'pd.to_datetime()', que converte diferentes representações de datas para um formato que o Pandas consegue manipular.

Isso facilita operações como:


*   ordenar datas
*   selecionar períodos
*   calcular diferenças
*   criar índices temporais



```python
import pandas as pd

datas = ["19/08/2026", "20/08/2026", "21/08/2026"]

datas_convertidas = pd.to_datetime(
    datas,
    format="%d/%m/%Y"
    )

datas_convertidas
```

Valores ausentes: NaT

Assim como o Pandas utiliza "NaN" para representar valores numéricos ausentes, ele utiliza "NaT" (*Not a Time*) para representar uma data ou horário ausente.

```python
datas = ["19/08/2026", None, "21/08/2026"]

datas_convertidas = pd.to_datetime(
    datas,
    format="%d/%m/%Y"
    )

datas_convertidas
```

Ao converter as datas para um formato reconhecido pelo Pandas, podemos utilizar recursos específicos para análise de séries temporais.

##DatetimeIndex

O "DatetimeIndex" é um índice de Pandas formado por datas e horários. Ao utilizar datas como índice de uma "Series" ou de um "DataFrame", o Pandas consegue reconhecer que os dados estão organizados temporalmente.

Isso permite realizar operações específicas com datas, como:


*   selecionar um determinado dia
*   selecionar um mês ou ano
*   filtrar intervalos de tempo
*   organizar e analisar os dados cronologicamente



```python
datas = pd.to_datetime([
    "2026-08-01",
    "2026-08-02",
    "2026-08-03",
    "2026-08-04",
    "2026-08-05"
])

vendas = [120, 135, 90, 128, 163]

serie = pd.Series(vendas, index=datas)

serie
```

```python
serie.index
```

Como nossas datas estão no índice, podemos utilizar as próprias datas para selecionar os dados.

```python
serie["2026-08-03"]
```

```python
serie["2026-08-02":"2026-08-04"]
```

Em resumo

O 'DatetimeIndex' permite que o Pandas trate as datas como parte fundamental dos dados. Com ele, podemos consultar e filtrar informações utilizando diretamente períodos de tempo.

##Criando uma Série Temporal

O Pandas possui a função 'pd.date_range' para gerar automaticamente uma sequência de datas.

Ela é útil quando as observações possuem uma frequência definida, como:


*   diariamente
*   semanalmente
*   mensalmente
*   anualmente



```python
datas = pd.date_range(
    start="2026-01-01",
    periods=12,
    freq="ME"
)

datas
```

Agora podemos associar um valor a cada uma das datas geradas.

```python
vendas = [120, 135, 90, 130, 200, 97, 150, 139, 187, 210, 180, 100]

serie_vendas = pd.Series(
    vendas,
    index=datas
)

serie_vendas
```

A frequência define de quanto em quanto tempo as observações acontecem.

```python
datas_diarias = pd.date_range(
    start="2026-01-01",
    periods=5,
    freq="D"
)

datas_diarias
```

```python
datas_semanais = pd.date_range(
    start="2026-01-01",
    periods=5,
    freq="W"
)

datas_semanais
```

Em resumo

A função 'pd.date_range()' permite criar sequências de datas de forma automática. O parâmetro 'freq' define a frequência das observações, tornando a criação de séries temporais mais prática.

#Manipulando uma Série Temporal

Quando as datas são utilizadas como índice, podemos selecionar facilmente:


*   uma data específica
*   um mês ou ano
*   um intervalo de tempo

Isso facilita a análise de diferentes períodos da série.



```python
serie_vendas["2026-03":"2026-06"]
```

```python
serie_vendas["2026"]
```

Além de selecionar períodos, podemos aplicar operações estatísticas aos valores da série como, por exemplo, calcular a média das vendas e descobrir o maior e menor valor.

```python
media_vendas = serie_vendas.mean()

media_vendas
```

```python
print("Maior valor:", serie_vendas.max())
print("Menor valor:", serie_vendas.min())
```

O Pandas permite selecionar períodos e aplicar operações estatísticas aos dados.

Dessa forma, podemos identificar valores máximos e mínimos, calcular médias e analisar o comportamento dos dados ao longo do tempo.

##Exemplo Prático

Vamos analisar as vendas mensais de uma loja durante o ano de 2026. Nesse exemplo, vamos:


*   Criar uma sequência de datas
*   Associar valores às datas
*   Criar uma série temporal
*   Selecionar um período
*   Realizar uma análise simples dos dados



```python
datas = pd.date_range(
    start="2026-01-01",
    periods=12,
    freq="ME"
)

datas
```

```python
vendas = [120, 135, 128, 150, 162, 170, 165, 180, 190, 185, 200, 215]

vendas
```

```python
serie_vendas = pd.Series(
    vendas,
    index=datas
)

serie_vendas
```

Analisando um período

Podemos selecionar apenas os meses que queremos analisar utilizando as próprias datas.

```python
periodo = serie_vendas["2026-03":"2026-06"]

periodo
```

```python
print("Média das vendas:", periodo.mean())
print("Maior número de vendas:", periodo.max())
print("Menor número de vendas:", periodo.min())
```
