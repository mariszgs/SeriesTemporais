##O que são séries temporais?
Uma série temporal é um conjunto de observações coletadas ao longo de um determinado tempo, sendo o tempo um fator fundamental, pois a ordem em que os dados são observados pode influenciar a análise.

###Exemplos


*   Temperatura registrada ao longo dos dias
*   Vendas de uma empresa ao longo dos meses
*   Inflação ao longo dos anos

As observações podem ocorrer em intervalos regulares ou irregulares
